```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import KMeans
```

```python
image_path = '../data/processed/cropped_player_1.jpg'
image = cv2.imread(image_path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(image)
plt.show()
```

```python
# top half of the image (since the jersey is usually on the top)
top_half = image[0: int(image.shape[0] / 2), :]
plt.imshow(top_half)
plt.show()
```

# Cluster the image into two clusters

```python
image_2d = top_half.reshape(-1, 3)

# perform k-means clustering with 2 clusters
kmeans = KMeans(n_clusters=2, random_state=0)
kmeans.fit(image_2d)

# get the cluster labels
labels = kmeans.labels_

# reshape the labels to the original image shape
clustered_image = labels.reshape(top_half.shape[0], top_half.shape[1])


plt.imshow(clustered_image)
plt.show()
```

```python
# cluster of the corners will be the background cluster (almost always) hopefully
corner_clusters = [clustered_image[0, 0], clustered_image[0, -1], clustered_image[-1, 0], clustered_image[-1, -1]]
non_player_cluster = max(set(corner_clusters), key=corner_clusters.count)
print(f'Non-player cluster: {non_player_cluster}')
```

```python
player_cluster = 1 - non_player_cluster  # since we have only 2 clusters, the other one is the player cluster
print(f'Player cluster: {player_cluster}')
```

```python
kmeans.cluster_centers_[player_cluster]
```
